## 1️⃣ Esplorazione del Dataset

**📂 File da utilizzare:** COVID_01-01-2021.csv

Creare un programma che:
- 👀 Visualizzi le **prime 5 righe** del DataFrame.
- 📊 Mostri le **informazioni generali** del dataset (numero di colonne, tipi di dato, numero di record).
- ⚠️ Controlli la presenza di **dati mancanti**.

```python
import pandas as pd

# Caricare il file CSV in un DataFrame
file_csv = "COVID_01-01-2021.csv"

try:
    # Leggere il file CSV
    df = pd.read_csv(file_csv)

    # Mostrare informazioni generali del DataFrame
    print("Informazioni del DataFrame:")
    print(df.info())

    # Calcolare e mostrare la quantità di dati mancanti per colonna
    print("\nQuantità di dati mancanti per colonna:")
    print(df.isnull().sum())

    # Mostrare le prime cinque righe del DataFrame
    print("\nPrime cinque righe del DataFrame:")
    print(df.head())

except FileNotFoundError:
    # Gestisce il caso in cui il file specificato non esiste nel percorso indicato
    # Questo può accadere se il nome del file è sbagliato o se il file è stato spostato
    print(f"Il file '{file_csv}' non è stato trovato. Assicurati che il file sia nella stessa cartella di questo script.")

except pd.errors.EmptyDataError:
    # Gestisce il caso in cui il file esiste ma è vuoto o non contiene dati leggibili
    # Aiuta a distinguere tra file mancanti e file esistenti ma vuoti
    print(f"Il file '{file_csv}' è vuoto o non contiene dati validi.")

except pd.errors.ParserError:
    # Gestisce errori di parsing, ad esempio quando il formato del file non è corretto
    # o quando ci sono problemi con i delimitatori o la struttura del CSV
    # ad esempio: delimitatori errati, virgolette non bilanciate, o formattazione non standard
    print(f"Si è verificato un problema durante l'analisi del file '{file_csv}'. Verifica che sia un file CSV valido.")

except Exception as e:
    # Cattura qualsiasi altro errore non previsto dalle eccezioni precedenti
    # Fornisce un messaggio generico con i dettagli dell'errore
    print(f"Si è verificato un errore imprevisto: {e}")
```

## 2️⃣ Analisi dei Dati per Paese

**📂 File da utilizzare:** COVID_01-01-2021.csv

Creare un programma che ottenga, per ciascun Paese, l'ultimo numero disponibile di:
- ✅ **Casi confermati**
- ❌ **Morti**
- 💚 **Recuperi**
- 🔵 **Casi attivi**

```python
try:
    df = pd.read_csv(file_csv)
    
    # Mostrare le colonne disponibili nel file
    print("Colonne del DataFrame:", df.columns.tolist())
    
    # Assicurarsi di utilizzare le colonne corrette
    colonne = ['Country_Region', 'Confirmed', 'Deaths', 'Recovered', 'Active']
    if all(col in df.columns for col in colonne):
        # Raggruppare i dati per paese e calcolare i totali
        totali_paese = df.groupby('Country_Region')[['Confirmed', 'Deaths', 'Recovered', 'Active']].sum()
        
        # Mostrare i totali
        print("Casi totali per paese:")
        print(totali_paese)
    else:
        print(f"Il file non contiene le colonne previste: {colonne}. Verifica i nomi.")
        
except FileNotFoundError:
    # Gestisce il caso in cui il file specificato non esiste nel percorso indicato
    # Questo può accadere se il nome del file è sbagliato o se il file è stato spostato
    print(f"Il file {file_csv} non è stato trovato. Assicurati che sia nel percorso specificato.")
    
except Exception as e:
    # Cattura qualsiasi altro errore non previsto dall'eccezione precedente
    # Ad esempio, problemi di permessi, errori di lettura del file, o errori durante l'elaborazione dei dati
    # Fornisce un messaggio generico con i dettagli dell'errore per facilitare il debug
    print(f"Si è verificato un errore: {e}")
```
