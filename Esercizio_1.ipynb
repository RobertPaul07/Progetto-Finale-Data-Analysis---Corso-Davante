📉 **Calcolare il numero di studenti che hanno fallito ciascuna materia**

```python
import numpy as np

nomi = np.array(["Robbie", "Marco", "Lore", "Fabri", "Alba"])
materie = np.array(["Latino", "Spagnolo", "Francese", "Inglese"])
voti = np.array([
    [8, 8, 9, 4],   # Robbie
    [7, 6, 7, 2],  # Marco
    [10, 7, 8, 9],  # Lore
    [4, 4, 3, 2],   # Fabri
    [9, 8, 9, 6]    # Alba
])

def mostra_bocciati(nomi, materie, voti):
    """
    Funzione che mostra il numero di studenti bocciati per ogni materia.
    Un voto inferiore a 6 è considerato insufficiente.
    
    Args:
        nomi: array con i nomi degli studenti
        materie: array con i nomi delle materie
        voti: matrice con i voti degli studenti per ogni materia

    Returns:
        bocciati_per_materia: il numero di bocciati per ogni materia
    """
    # Converte l'input in un array numpy se non lo è già
    voti_array = np.array(voti)
    
    # Itera attraverso ogni materia (colonne della matrice)
    for j, materia in enumerate(materie):
        # Conta quanti studenti hanno un voto insufficiente (<6) nella materia corrente
        # voti_array[:, j] seleziona tutti i voti della colonna j (materia corrente)
        bocciati = np.sum(voti_array[:, j] < 6)
        
        # Stampa il risultato con un messaggio appropriato (singolare o plurale)
        if bocciati > 1:
            print(f"In {materia} sono stati bocciati {bocciati} studenti.")
        else:
            print(f"In {materia} è stato bocciato {bocciati} studente.")

mostra_bocciati(nomi, materie, voti)
```

❌ **Calcolare il numero di studenti che hanno fallito in ogni materia**

```python
import numpy as np

def bocciati_tutto(nomi: np.ndarray, voti: np.ndarray) -> np.ndarray:
    """
    Restituisce gli studenti che hanno tutti i voti inferiori a 6
    utilizzando loop nidificati.

    Args:
    nomi : np.ndarray
        Array dei nomi degli studenti.
    voti : np.ndarray
        Matrice dei voti (righe = studenti, colonne = materie).

    Returns:
    np.ndarray
        Array contenente i nomi degli studenti bocciati in tutte le materie.
    """

    bocciati = []  # Lista temporanea

    for i in range(len(nomi)):  # Ciclo sugli studenti
        bocciato = True

        for j in range(len(voti[i])):  # Ciclo sui voti dello studente
            if voti[i][j] >= 6:
                bocciato = False
                break  # Esce dal ciclo interno

        if bocciato:
            bocciati.append(nomi[i])

    return np.array(bocciati)

    # UTILIZZO
if __name__ == "__main__":

    nomi = np.array(["Robbie", "Marco", "Lore", "Fabri", "Alba"])

    voti = np.array([
        [8, 8, 9, 4],
        [7, 6, 7, 2],
        [10, 7, 8, 9],
        [4, 4, 3, 2],
        [9, 8, 9, 6]
    ])

    bocciati = bocciati_tutto(nomi, voti)

    if bocciati.size > 0:
        print("Studenti bocciati in tutte le materie:\n")
        print(bocciati)
    else:
        print("Nessuno studente è bocciato in tutte le materie.")
```

📊 **Calcolare la media dei voti di ogni studente**

```python
def calcolare_media(nomi, voti):
    """
    Calcola e visualizza le medie dei voti per ogni studente e per l'intera classe.
    
    Args:
        nomi: array con i nomi degli studenti
        voti: matrice con i voti degli studenti (righe=studenti, colonne=materie)

    Returns:
        medie_studenti, media_classe: liste delle medie individuali e media complessiva    
    """
    # Calcola la media per ogni studente lungo l'asse delle colonne (axis=1)
    medie = np.mean(voti, axis=1)
    
    # Intestazione per il riepilogo
    print("Riepilogo delle medie:")
    print("-" * 60)  # Linea separatrice per migliorare la leggibilità
    
    # Itera su ogni studente e mostra la sua media
    for i, nome in enumerate(nomi):
        media = medie[i]
        print(f"{nome} ha ottenuto una media di {media:.2f}.")  # Formatta con 2 decimali
    
    # Calcola e mostra la media dell'intera classe
    media_classe = np.mean(medie)  # Media delle medie individuali
    print(f"\nLa media della classe è {media_classe:.2f}.")
    

calcolare_media(nomi, voti)
```

🏆 **Individuare gli studenti con media superiore a 5**

```python
medie = np.mean(voti, axis=1)

# Identificazione degli studenti con media superiore a 5
studenti_sopra_5 = medie > 5

# Visualizzazione dei risultati
print("Studenti con voto medio superiore a 5:")
print("-" * 50)

# Itera attraverso gli studenti e mostra solo quelli con media > 5
for i, nome in enumerate(nomi):
    if studenti_sopra_5[i]:  # Controlla se lo studente ha media superiore a 5
        print(f"{nome}: media {medie[i]:.2f}")  # Formatta la media con 2 decimali

# Statistiche aggiuntive
num_sopra_5 = np.sum(studenti_sopra_5)
print(f"\nTotale: {num_sopra_5} studenti su {len(nomi)} hanno una media superiore a 5")
```

# 📊🎓 Calcolo degli Studenti Promossi

Di seguito è riportata una semplice parte di codice aggiuntiva, ideata per:

✅ Calcolare il numero di **studenti promossi**

📈 Determinare la **percentuale di successo**

🖥️ Visualizzare i **risultati a schermo**

```python
# Una semplice parte di codice aggiuntiva, da me ideata per calcolare e mandare a schermo gli
# studenti promossi e la percentuale di successo

def calcolare_promossi(nomi, voti):
    """
    Determina quali studenti sono promossi in base ai loro voti e visualizza i risultati.
    Uno studente è considerato promosso solo se ha ottenuto un voto sufficiente (≥6) in tutte le materie.
    
    Args:
        nomi: array con i nomi degli studenti
        voti: matrice con i voti degli studenti (righe=studenti, colonne=materie)

    Returns:
        promossi: array booleano dove True indica che lo studente è promosso
    """
    # Crea un array booleano dove True indica che lo studente ha tutti i voti sufficienti
    # np.all(condizione, axis=1) verifica che la condizione sia vera per tutti gli elementi di ogni riga
    promossi = np.all(voti >= 6, axis=1)
    
    # Calcola il numero di studenti promossi e il totale degli studenti
    num_promossi = np.sum(promossi)  
    num_studenti = len(nomi)        
    
    # Intestazione con il riepilogo dei risultati
    print(f"Risultati finali del corso ({num_promossi}/{num_studenti} promossi):")
    print("-" * 40)  # Linea separatrice per migliorare la leggibilità
    
    # Elenca gli studenti promossi
    print("Gli studenti che hanno superato il corso sono:")
    for i, promosso in enumerate(promossi):
        if promosso: 
            print(f"- {nomi[i]}")
    
    # Elenca gli studenti non promossi (solo se ce ne sono)
    if num_studenti > num_promossi:  # Verifica se ci sono studenti non promossi
        print("\nGli studenti che non hanno superato il corso sono:")
        for i, promosso in enumerate(promossi):
            if not promosso: 
                print(f"- {nomi[i]}")
    
    # Calcola e visualizza la percentuale di successo del corso
    percentuale = (num_promossi / num_studenti) * 100
    print(f"\nPercentuale di successo: {percentuale:.1f}%")  # Formatta con 1 decimale

calcolare_promossi(nomi, voti)
```
