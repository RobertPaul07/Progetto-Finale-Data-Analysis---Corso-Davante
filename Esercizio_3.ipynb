## 1️⃣🔢   Creazione di Serie

Creare due serie:

- **Serie B** → a partire da un array di numeri generato con **NumPy**
- **Serie A** → a partire da una lista di numeri

```python
import numpy as np
import pandas as pd

# Generare una lista di numeri interi casuali tra 0 e 20 e convertirla in una serie
lista_numeri = np.random.randint(0, 21, 10).tolist()
serie_A = pd.Series(lista_numeri, name="serieA")
print("Contenuto della serie A:")
print(serie_A)

# Generare un array di 10 numeri interi casuali tra 0 e 20 e convertirlo in una serie
array_numeri = np.random.randint(0, 21, 10)
serie_B = pd.Series(array_numeri, name="serieB")
print("\nContenuto della serie B:")
print(serie_B)
```

## 2️⃣📍    Multipli di 3

Individuare la **posizione** degli elementi della **Serie B** che sono **multipli di 3**.

```python
def trovare_posizioni_multiplo_3(serie_B):
    """
    Trova gli indici dei valori multipli di 3 in una pandas Series.
    
    Args:
        serie_B: pandas Series contenente valori numerici

    Returns:
        pandas.Index: Gli indici dei valori che sono multipli di 3    
    """
    # Crea una maschera booleana per identificare i valori divisibili per 3
    # e seleziona solo gli indici di questi elementi
    indici = serie_B[serie_B % 3 == 0].index
    
    # Stampa un'intestazione per chiarire l'output
    print("\nPosizioni dei numeri multipli di 3 nella serie:")
    
    # Converte l'oggetto Index di pandas in una lista Python e lo stampa
    # Output più leggibile e facilmente utilizzabile
    print(list(indici))

trovare_posizioni_multiplo_3(serie_B)
```

## 3️⃣📋    Creazione di un DataFrame

Creare un **DataFrame** combinando le due serie **A** e **B** e assegnare un **nome all'indice**.

```python
# Crea un nuovo DataFrame combinando le due serie
# serie_A e serie_B diventano colonne del DataFrame con i nomi 'serieA' e 'serieB'
df = pd.DataFrame({
    'serieA': serie_A,
    'serieB': serie_B
})

# Creazione di indici personalizzati
indici_personalizzati = [f"InD_{i+1}" for i in range(len(df))]

# Assegnazione degli indici
df.index = indici_personalizzati

# Assegna un nome all'indice
df.index.name = "New Index"

# Visualizzazione del DataFrame con formattazione migliorata
print("\n" + "="*50)
print("DataFrame combinato dalle serie A e B:")
print("="*50)
print(df.head(10))

# Statistiche descrittive
print("\n" + "="*50)
print("Statistiche descrittive:")
print("="*50)
print(df.describe())
```

## 4️⃣🔄    Elementi Comuni

Individuare gli **elementi comuni** presenti sia nella **Serie A** sia nella **Serie B**.

```python
def elementi_comuni(serie_A, serie_B):
    """
    Trova gli elementi comuni tra due pandas Series.
    
    Args:
        serie_A (pandas.Series): Prima serie di elementi
        serie_B (pandas.Series): Seconda serie di elementi
        
    Returns:
        pandas.Series: Una nuova Series contenente solo gli elementi di serie_A
                      che sono presenti anche in serie_B, mantenendo gli indici
                      originali di serie_A
    """
    # Utilizza il metodo isin() per creare una maschera booleana
    # che identifica gli elementi di serie_A presenti anche in serie_B
    # Poi filtra serie_A mantenendo solo questi elementi comuni
    comuni = serie_A[serie_A.isin(serie_B)]
    
    # Stampa un'intestazione per chiarire il contenuto dell'output
    print("\nElementi appartenti a entrambe le serie:")
    
    # Stampa la serie filtrata contenente solo gli elementi comuni
    # Questo mostrerà sia i valori che i loro indici originali in serie_A
    print(comuni)

elementi_comuni(serie_A, serie_B)
```

## 5️⃣➖    Differenza tra le Serie

Individuare gli elementi della **Serie A** che **non sono presenti** nella **Serie B**.

```python
def elementi_non_comuni(serie_A, serie_B):
    """
    Trova gli elementi di serie_A che non sono presenti in serie_B.
    
    Args:
        serie_A (pandas.Series): Prima serie di elementi
        serie_B (pandas.Series): Seconda serie di elementi
        
    Returns:
        pandas.Series: Una nuova Series contenente solo gli elementi di serie_A
                       che NON sono presenti in serie_B, mantenendo gli indici
                       originali di serie_A
    """
    # Utilizza il metodo isin() per identificare gli elementi di serie_A presenti in serie_B
    # Il simbolo ~ inverte la maschera booleana, selezionando gli elementi NON presenti
    # serie_A[~serie_A.isin(serie_B)] filtra serie_A mantenendo solo gli elementi esclusivi
    non_comuni = serie_A[~serie_A.isin(serie_B)]
    
    # Stampa un'intestazione descrittiva per l'output
    print("\nElementi della prima serie che non sono presenti nella seconda:")
    
    # Stampa la serie filtrata contenente solo gli elementi esclusivi di serie_A
    # Questo mostrerà sia i valori che i loro indici originali
    print(non_comuni)

elementi_non_comuni(serie_A, serie_B)
```

## 6️⃣📐    Conversione in DataFrame

Convertire la seguente struttura in un **DataFrame** di:

- **7 righe**
- **5 colonne**

Partendo dalla seguente serie:

`series = pd.Series(np.random.randint(1, 10, 35))`

```python
# Creare una serie con 35 numeri interi casuali tra 1 e 9
series = pd.Series(np.random.randint(1, 10, 35))

# Convertire la serie in un DataFrame di 7 righe e 5 colonne
# Utilizzo il metodo reshape per riorganizzare i dati
matrice = series.values.reshape(7, 5)
df = pd.DataFrame(matrice, columns=['Col.1', 'Col.2', 'Col.3', 'Col.4', 'Col.5'])

# Visualizzare il DataFrame risultante
print("\nDataFrame di 7 righe e 5 colonne estrapolato da una serie:")
print("-" * 60)
print(df)
```

# Analisi di un DataFrame Multivariato

## 🎯 Obiettivo
Creare un programma che generi un DataFrame Pandas con dati numerici simulati (punteggi di studenti in più materie) e calcoli statistiche descrittive come media e deviazione standard per colonna.

##  💻  Il programma:

#### 🎲 Genera dati numerici simulati (punteggi studenti)

#### 📋 Crea un DataFrame Pandas

#### 📊 Calcola la media per colonna

#### 📉 Calcola la deviazione standard per colonna

```python
import numpy as np
import pandas as pd


def genera_dataframe_studenti(num_studenti: int = 10) -> pd.DataFrame:
    """
    Genera un DataFrame contenente punteggi simulati di studenti in diverse materie.
    Ho optato per punteggi e non voti da 1 a 10 per avere un range d'azione più ampio.

    Args:
    num_studenti : int
        Numero di studenti da generare.

    Returns:
    pd.DataFrame
        DataFrame con punteggi simulati.
    """

    np.random.seed(42)  # Per riproducibilità

    dati = {
        "Matematica": np.random.randint(18, 31, num_studenti),
        "Fisica": np.random.randint(18, 31, num_studenti),
        "Informatica": np.random.randint(18, 31, num_studenti),
        "Statistica": np.random.randint(18, 31, num_studenti),
    }

    df = pd.DataFrame(dati)
    df.index = [f"Studente_{i+1}" for i in range(num_studenti)]

    return df


def calcola_statistiche(df: pd.DataFrame) -> tuple:
    """
    Calcola media e deviazione standard per ciascuna colonna numerica.

    Args:
    df : pd.DataFrame

    Returns:
    tuple
         (medie, deviazioni_standard)
    """

    colonne_numeriche = df.select_dtypes(include=np.number)

    medie = colonne_numeriche.mean()
    deviazioni_standard = colonne_numeriche.std()

    return medie, deviazioni_standard


    # UTILIZZO

if __name__ == "__main__":

    df_studenti = genera_dataframe_studenti(10)

    print("📋 DataFrame generato:\n")
    print(df_studenti)

    medie, deviazioni = calcola_statistiche(df_studenti)

    print("\n📊 Media per colonna:\n")
    print(medie)

    print("\n📉 Deviazione standard per colonna:\n")
    print(deviazioni)
```
