<h2> 3️⃣ Paesi senza pazienti recuperati</h2>

<p><b>📂 File da utilizzare:</b> COVID_01-01-2021.csv</p>

<ul>
    <li>Creare un programma che ottenga una <b>lista di paesi</b> dove non ci sono casi di pazienti recuperati.</li>
    <li>🔍 Analizzare i dati per identificare rapidamente i paesi con <b>0 pazienti guariti</b>.</li>
</ul>

```python
import pandas as pd

# Caricare il file CSV in un DataFrame
file_csv = "COVID_01-01-2021.csv"

try:
    df = pd.read_csv(file_csv)
    
    # Mostrare le colonne disponibili nel file
    print("Colonne del DataFrame:", df.columns.tolist())
    
    # Assicurarsi che le colonne siano corrette
    colonne = ['Country_Region', 'Province_State', 'Recovered']  # Aggiornate alle colonne reali
    if all(col in df.columns for col in colonne):
        # Filtrare dove 'Recovered' è 0 o mancante (NaN)
        senza_recuperati = df[(df['Recovered'].isnull()) | (df['Recovered'] == 0)]
        
        # Selezionare solo le colonne 'Country_Region' e 'Province_State'
        risultato = senza_recuperati[['Country_Region', 'Province_State']]
        
        # Rimuovere eventuali duplicati
        risultato = risultato.drop_duplicates()
        
        # Mostrare la lista di province e paesi senza casi recuperati
        print("Province e paesi senza casi recuperati:")
        print(risultato)
    else:
        print(f"Il file non contiene le colonne previste: {colonne}. Verifica i nomi.")
        
except FileNotFoundError:
    # Gestisce il caso in cui il file specificato non esiste nel percorso indicato
    # Questo può accadere se il nome del file è sbagliato o se il file è stato spostato
    print(f"Il file {file_csv} non è stato trovato. Assicurati che sia nel percorso specificato.")
    
except Exception as e:
    # Cattura qualsiasi altro errore non previsto dall'eccezione precedente
    # Ad esempio, problemi di permessi, errori di lettura del file, o errori durante l'elaborazione dei dati
    # Fornisce un messaggio generico con i dettagli dell'errore per facilitare il debug
    print(f"Si è verificato un errore: {e}")
```

<h2> 4️⃣ Top 10 Paesi per casi confermati</h2>

<p><b>📂 File da utilizzare:</b> COVID_01-01-2021.csv</p>

<ul>
    <li>Creare un programma che ottenga, per i <b>10 paesi con il maggior numero di casi confermati</b>:
        <ul>
            <li>🌍 <b>Paese</b></li>
            <li>✅ <b>Casi confermati</b></li>
            <li>❌ <b>Morti</b></li>
            <li>💚 <b>Pazienti recuperati</b></li>
        </ul>
    </li>

```python
try:
    df = pd.read_csv(file_csv)

    # Mostrare le colonne del file per verifica
    print("Colonne disponibili nel file:", df.columns.tolist())

    # Assicurarsi che le colonne siano corrette
    colonne = ['Country_Region', 'Confirmed', 'Deaths', 'Recovered', 'Active']  # Aggiornate alle colonne reali
    if all(col in df.columns for col in colonne):
        # Raggruppare i dati per paese e calcolare i totali
        totali_paese = df.groupby('Country_Region')[['Confirmed', 'Deaths', 'Recovered', 'Active']].sum()

        # Ordinare per casi confermati in ordine decrescente
        top_10_paesi = totali_paese.sort_values(by='Confirmed', ascending=False).head(10)

        # Mostrare i 10 paesi con più casi confermati
        print("Top 10 paesi con più casi confermati:")
        print(top_10_paesi)
    else:
        print(f"Il file non contiene le colonne previste: {colonne}. Verifica i nomi delle colonne.")
        
except FileNotFoundError:
    # Gestisce il caso in cui il file specificato non esiste nel percorso indicato
    # Questo può accadere se il nome del file è sbagliato o se il file è stato spostato
    print(f"Il file '{file_csv}' non è stato trovato. Assicurati che sia nella stessa cartella di questo script.")
    
except pd.errors.EmptyDataError:
    # Gestisce il caso in cui il file esiste ma è vuoto o non contiene dati leggibili
    # Aiuta a distinguere tra file mancanti e file esistenti ma vuoti
    print(f"Il file '{file_csv}' è vuoto o non contiene dati validi.")
    
except pd.errors.ParserError:
    # Gestisce errori di parsing, ad esempio quando il formato del file non è corretto
    # o quando ci sono problemi con i delimitatori o la struttura del CSV
    # ad esempio: delimitatori errati, virgolette non bilanciate, o formattazione non standard
    print(f"Si è verificato un problema nell'analisi del file '{file_csv}'. Verifica che sia un CSV valido.")
    
except Exception as e:
    # Cattura qualsiasi altro errore non previsto dall'eccezione precedente
    # Ad esempio, problemi di permessi, errori di lettura del file, o errori durante l'elaborazione dei dati
    # Fornisce un messaggio generico con i dettagli dell'errore per facilitare il debug:
    print(f"Si è verificato un errore imprevisto: {e}")
```
